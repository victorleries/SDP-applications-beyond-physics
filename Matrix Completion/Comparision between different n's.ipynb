```python
#import necessary libraries
import numpy as np
import cvxpy as cp
import matplotlib.pyplot as plt
```

```python
DIM=[3,10,15,20] #does the plots for n=2, 10, 15 and 20
font={'family':'times new roman'}
for n in DIM:
    N=str(n)
    A=np.random.rand(n)
    A=np.outer(A,A) #generate a matrix of dimension n and rank 1
    d=[]
    M=[]
    for m in range(1,n**2): #generate omega set
        M.append(m)
        om=[]
        om.append(np.random.randint(0,n,size=2))
        while (len(om)!=m):
            a=[]
            l=np.random.randint(0,n,2)
            for j in range(len(om)):
                if(l!=om[j]).any():
                    a.append(True)
                elif(l==om[j]).all():
                    a.append(False)
            a=np.array(a)
            if a.all():
                om.append(l)
        om=np.array(om)
        #solve SDP
        X = cp.Variable((n,n))

        obj = cp.norm(X,"nuc")

        constraints = []
        for i in range(m):
            constraints += [X[om[i,0],om[i,1]] == A[om[i,0],om[i,1]]]

        p = cp.Problem(cp.Minimize(obj),constraints)
        p.solve()
        S=np.array(X.value)
        A=np.array(A)
        d.append(np.linalg.norm(S-A,'nuc')) #calculate the distance
    d=d/np.linalg.norm(A,'nuc')
    plt.plot(M,d,'.') #plot data
    plt.xlabel("Entries known",fontdict=font)
    plt.ylabel("Distance between completed and original matrix",fontdict=font)
    plt.savefig(N)
    plt.clf()
```
