```python
#import necessary libraries
import numpy as np
import cvxpy as cp
import matplotlib.pyplot as plt
```

```python
#definie tha parameters of the problem
n=18
m=150
d=[]
K=[]
A=np.random.rand(n)
A=np.outer(A,A) #generate M0
```

```python
for k in range(n):#go from range 1 to 18
    K.append(k+1)
    om=[] #generate Omega set
    om.append(np.random.randint(0,n,size=2))
    while (len(om)!=m):
        a=[]
        l=np.random.randint(0,n,2)
        for j in range(len(om)):
            if(l!=om[j]).any():
                a.append(True)
            elif(l==om[j]).all():
                a.append(False)
        a=np.array(a)
        if a.all():
            om.append(l)
    om=np.array(om)
    #solve SDP
    X = cp.Variable((n,n))

    obj = cp.norm(X,"nuc")

    constraints = []
    for i in range(m):
        constraints += [X[om[i,0],om[i,1]] == A[om[i,0],om[i,1]]]

    p = cp.Problem(cp.Minimize(obj),constraints)
    p.solve()
    S=np.array(X.value) #save date to plot
    A=np.array(A)
    d.append(np.linalg.norm(S-A,'nuc')/np.linalg.norm(A,'nuc')) #calculate distance d
    B=np.random.rand(n)
    B=np.outer(B,B)
    A+=B #generate Mi
```

```python
#generate plot
font={'family':'times new roman'}
ti=[]
for i in range(1,19):
    ti.append(i)
plt.plot(K,d,'.')
plt.xlabel("Matrix rank",fontdict=font)
plt.xticks(ti)
plt.yticks([0,0.05,0.1,0.15,0.2,0.25,0.3])
plt.ylabel("Distance between completed and original matrix",fontdict=font)
plt.savefig("How big can the range be")
plt.clf()
```
